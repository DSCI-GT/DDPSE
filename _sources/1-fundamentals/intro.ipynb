# Optimization Fundamentals

This module sets up the language of optimization and the tools we use for the rest of the book. Topic 1.1 shows how an engineering decision becomes a formulation and why the class of a formulation (linear or nonlinear, continuous or integer, convex or not) decides which solvers we can trust. Topic 1.2 works one example of each class. Topic 1.3 introduces Pyomo, the modeling language we use from here on, and Topics 1.4 and 1.5 treat linear programming and the simplex method.

The module also introduces the book's running example: water electrolysis, first as a chemical equilibrium set by the applied voltage, then as a plant of electrolyzer stacks that must meet a hydrogen demand at least cost. The first view fixes the price of hydrogen. The second asks which stacks should do the work, and it returns, with one new complication at a time, in every topic through Topic 3.6.

Lecture weeks 1–3 (Modules 1–3).
