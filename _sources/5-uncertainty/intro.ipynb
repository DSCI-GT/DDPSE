# Optimization under Uncertainty

Every optimization problem in this book so far has treated its parameters as known. In practice, prices, demands and measured properties are uncertain, and a decision that is optimal for one value of a parameter can be poor for another. This bonus module introduces the two main ways to plan for that: stochastic programming, which optimizes the expected outcome over a set of scenarios (Topic 5.1), and robust optimization, which protects against the worst case within a set of possible values (Topic 5.2).
